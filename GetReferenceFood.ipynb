### **This is the code to get the reference food for a participant who has already been run through the health and taste ratings of the Food Choice Task. Please select the suitable code to run (only one out of the two subsections!). Before running, please upload the ppt's data file(s) using the left toolbar.**

#### A) Use the following code cell if the health and taste ratings for the ppt are in the same data file. Please replace ABC with the data file name. Example: datafile = pandas.read_csv('861690_FCT_2026-04-07_13h36.06.985.csv')

```python
import pandas
datafile = pandas.read_csv('861690_FCT_2026-04-07_13h36.06.985.csv')  #Replace the file name
health_df = datafile[(datafile['health_rating'].notna()) & (datafile['health_rating']!='')]
health_df = health_df[['health_rating', 'food_item']]
taste_df = datafile[(datafile['taste_rating'].notna()) & (datafile['taste_rating']!='')]
taste_df = taste_df[['taste_rating', 'food_item']]
merged = health_df.merge(taste_df,how="inner",left_on="food_item",right_on="food_item")
ref_food_df = merged[((merged['health_rating'] == 3.0) & (merged['taste_rating'] == 3.0))]
if len(ref_food_df) == 0:
  ref_food_df = merged[((merged['health_rating'] == 3.0) & (merged['taste_rating'] == 4.0)) | ((merged['health_rating'] == 4.0) & (merged['taste_rating'] == 3.0))]
if len(ref_food_df) == 0:
  ref_food_df = merged[((merged['health_rating'] == 3.0) & (merged['taste_rating'] == 2.0)) | ((merged['health_rating'] == 2.0) & (merged['taste_rating'] == 3.0))]
ref_food_df.reset_index(inplace=True)
if len(ref_food_df) == 0:
  print("No reference food found. Please don't enter any reference food while running the task and it will use the backup option of saltines.")
else:
  print("Reference food found is: "+ref_food_df.loc[0,'food_item'])
  print("Please copy paste the exact food name including spaces in the URL parameter, and please make sure you don't include any space before or after the food name.")
```
